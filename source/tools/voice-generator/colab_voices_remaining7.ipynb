(Sayed already uses Karim; every voice here is different from his and from each other.) I cannot listen, so if a voice sounds wrong for the character, change the name in the next cell and re-run that character's cell.

```
import subprocess, json
VOICES = {"noga": "Rana", "shawky": "Youssef", "hankash": "Mahmoud", "kabsha": "Shaker", "asaliya": "Hanan", "faheem": "Hassan", "boga": "Ramy"}
r = subprocess.run(["voicetut", "--list-speakers"], capture_output=True, text=True)
names = [x.strip(" -*\t") for x in (r.stdout + r.stderr).splitlines() if x.strip()]
print("Available voices:", ", ".join(names))
low = {n.lower(): n for n in names}
bad = {c: v for c, v in VOICES.items() if v.lower() not in low}
if bad: print("STOP: these voices are not in the list above:", bad, "- edit VOICES and run this cell again."); raise SystemExit
if len({v.lower() for v in VOICES.values()}) != len(VOICES) or "karim" in {v.lower() for v in VOICES.values()}:
    print("STOP: two characters share a voice (or use Sayed's Karim)."); raise SystemExit
sp = {c: "" for c in ["sayed","noga","shawky","hankash","boga","kabsha","asaliya","faheem"]}
sp.update({c: low[v.lower()] for c, v in VOICES.items()})
json.dump(sp, open("/content/arees/tools/voice-generator/speakers.json", "w"))
print("OK voice map:", {c: low[v.lower()] for c, v in VOICES.items()})
```

```
import subprocess, os, json, torch, gc, soundfile as sf, numpy as np
from IPython.display import Audio, display
ROOT = "/content/arees"
def make_and_check(char):
    # 1) generate this character only (separate process => GPU memory is released when it ends)
    p = subprocess.Popen(["python", ROOT + "/tools/voice-generator/generate.py", "--only", char, "--force"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    out = p.stdout.read(); p.wait()
    if p.returncode != 0: print(out[-2500:]); print("STOP: generator failed for", char); return False
    torch.cuda.empty_cache(); gc.collect()
    # 2) validate every line of this character, then play
    lines = [l for l in json.loads(LINES) if l["char"] == char]
    man = json.load(open(ROOT + "/assets/audio/voices/manifest.json", encoding="utf-8")).get(char, {})
    bad = []
    for l in lines:
        path = ROOT + "/" + man.get(l["key"], "missing")
        if not os.path.isfile(path): bad.append(l["key"] + " missing"); continue
        d, sr = sf.read(path); dur = len(d) / sr; rms = float(np.sqrt(np.mean(np.square(d)))); pk = float(np.max(np.abs(d))); cps = len(l["text"]) / max(dur, 0.01)
        ok = os.path.getsize(path) > 8000 and dur > 0.7 and rms > 0.005 and pk < 0.999 and 3 < cps < 30
        print(("OK  " if ok else "BAD ") + l["key"].ljust(10), f"{dur:4.1f}s {sr}Hz loud {rms:.3f} peak {pk:.2f} {cps:.0f} chars/s |", l["text"])
        if not ok: bad.append(l["key"] + " bad audio")
        display(Audio(path))
    if bad: print("STOP: problems for", char, bad); return False
    print("OK", char, "-", len(lines), "files valid. Now press the players above and LISTEN."); return True
print("helper ready")
```
